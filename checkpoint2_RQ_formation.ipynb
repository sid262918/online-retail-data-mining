```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

online_retail = pd.read_excel("/content/drive/MyDrive/colab_notebooks/Data_mining/Online Retail.xlsx")
online_retail.head()
```

This dataset has 541909 transaction records and 8 variables. Each row represents a product within a transaction. Most of the variables are complete, except for Description and CustomerID, which has missing values. There are alos negative values in the Quantity variable and UnitPrice. This implies that some of the items could have been returned. This suggests a need for data cleaning before conducting the itemset mining. The span of the dataset is from December 2010 to December 2011 and has Transactional, Temporal, and information on pricing, making it a good for pattern mining analysis.

```python
# remove cancelled invoices
online_retail = online_retail[~online_retail['InvoiceNo'].astype(str).str.startswith('C')]

# remove -/0
online_retail = online_retail[online_retail['Quantity'] > 0]

# remove -/0 prices
online_retail = online_retail[online_retail['UnitPrice'] > 0]

online_retail.shape
```

After removing the cancelled invoices, negative values, and invalid prices, the dataset was reduced 530104 rows. This ensures that only completed purchases are analyzed.

```python
basket_size = online_retail.groupby('InvoiceNo')['StockCode'].count()
print(basket_size.describe())
```

## Checkpoint 2 - Research Questions Formation

# Project Scope
For this project, Online Retail dataset was used. This dataset contains transactional records from a United Kingdom based online retailer. The Online Retail dataset is ideal for basket-based analysis because each of the transactions include several products.

<br> From the EDA we have significant patterns that were observed. Most of the transactions contain a small number of items and a few of the transactions contain large quantities, causing the basket sizes to be highly skewed. The product frequency also shows a long-tail distribution, in which a small nummber of the products appear often while most of the products appear rarely. last but not least, the transaction volume indicates temporal fluctuation, with signs of it increasing towards the end of the year.

<br> These characteristics show that the Online Retail dataset is suitable for frequent itemset mining and association rule mining (course techniques). With the presence of timestamp data, sequential pattern (external technique) mining can be used as well.



# Research Question Definition
RQ1: What frequent itemset appear in the Online Retail dataset when varying the support threshold?
* *Data mining task type:* Frequent itemset mining
* *Relevant algorithms:* Apriori (course), FP-Growth
* *Evaluation Criteria:* Support and number of itemset

<br> RQ2: How do the confidence and lift compare when dealing with association rules evaluated from the retail transactions?
* *Data mining task type:* Association Rule mining
* *Relevant algorithms:* Association Rule generation
* *Evaluation Criteria:* Confidence, lift, and support

<br> RQ3: Can Sequntial patterns show purchasing relationships that would not be captured by unordered frequent itemsets?
* *Data mining task type:* sequential pattern mining
* *Relevant algorithms:* PrefixSpan (external)
* *Evaluation Criteria:* Pattern support and interpretability

# RQ-to-Method Mapping
RQ1 focuses on frequent itemset mining and will find products that appear more often using methods like Apriori and FP-Growth. The evaluation metric will be support.

<br> RQ2 focuses on association rule mining. This will check the strength and usefulness of the rules received from frequent itemsets using confidence and lift.

<br> RQ3 uses an algorithm like PrefixSpan for the sequential pattern mining. The purpose is to determine whether having the transaction order shows any new patterns that are not captured by frequent itemset mining.

# Motivation and Feasibility
The EDA showed many significant characteristics of the dataset, such as long-tail distribution of products, large variation in basket sizes, and clear temporal patterns in the transaction activity. These results encourage the usage of both techniques learned in the course and found externally.

<br> Frequent itemset mining is the optimal choice because the dataset is organized as transactions with many items. This is extended by association rule mining, which checks the strength of the relationships between the products. However, These methods treat the transactions as unordered sets, which implies they dont capture temporal ordering of the purchases.

<br> Due to this limitation, it is wise to use the sequential pattern mining as an external technique. By utilizing the time information, sequential methods may reveal patterns that are not visible to association rules.

<br> The dataset has already been cleaned and arranged into transactions, making the project feasible. Standard Python libraries support FP-Growth and association rule generation, and sequential pattern mining can be used by utilizing existing tools. Some risks include sensitivity to supporth thresholds and the computational cost for sequential method. These can be dealt with and controlled with careful experimentation and parameter tuning.

<br>This problem is non-trivial because frequent itemset methods ignore ordering, which may cause missing significant temporal relationships in the data.

# Methodological Planning
The project will start with frequent itemset mining using FP-Growth, and Apriori will be used as a baseline method for comparison. These methods will find product combinations that are often used in transactions. FP-Growth would be preffered because it is more efficient than Apriori for larger datasets, as it does not generate a huge number of candidate itemsets.

<br> The frequent itemsets will then be used to generate association rules. To assess the quality of these rules, Support, Confidence, and Lift will be used. These metrics will help figure out if the often occuring rules are meaningful and informative.

<br> Lastly, sequential pattern mining is used to examine the order of purchases over time. This step will help in figuring out whether utilizing the temporal data reveals more structure in customer purchasing behavior.

<br> Inital testing will focus on confirming that the dataset is correctly formatted for each method and that the libraries function as intended. Before scaling up to the entire dataset, subsets of the data will be used to check computational feasibility.
